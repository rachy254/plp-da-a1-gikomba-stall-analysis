## define_dataset_and_functions

### Subtask:
Initialize the raw dataset and implement the core Python functions for analysis without using pandas or numpy.


**Reasoning**:
I will create a Python code block to define the raw vegetable stall dataset and implement the requested pure Python functions (`week_totals`, `classify_day`, and a custom bubble sort function) without importing pandas or numpy as instructed.



```python
# 1. Define the raw vegetable stall dataset as a list of dictionaries
wanjiku_dataset = [
    {'day': 'Monday', 'item': 'Kale (Sukuma Wiki)', 'quantity': 50, 'price_per_unit': 40},
    {'day': 'Tuesday', 'item': 'Spinach', 'quantity': 35, 'price_per_unit': 50},
    {'day': 'Wednesday', 'item': 'Cabbage', 'quantity': 20, 'price_per_unit': 80},
    {'day': 'Thursday', 'item': 'Tomatoes', 'quantity': 60, 'price_per_unit': 60},
    {'day': 'Friday', 'item': 'Onions', 'quantity': 45, 'price_per_unit': 70},
    {'day': 'Saturday', 'item': 'Potatoes', 'quantity': 80, 'price_per_unit': 50},
    {'day': 'Sunday', 'item': 'Carrots', 'quantity': 25, 'price_per_unit': 90}
]

# 2. Implement week_totals(dataset)
def week_totals(dataset):
    """
    Calculates and returns the total sales revenue for the week.

    Parameters:
    dataset (list of dict): The list of dictionaries containing daily sales data.

    Returns:
    float: The total sales revenue for the week.
    """
    total_revenue = 0.0
    for record in dataset:
        total_revenue += record['quantity'] * record['price_per_unit']
    return total_revenue

# 3. Implement classify_day(sales_amount)
def classify_day(sales_amount):
    """
    Categorizes a day's sales as 'High', 'Medium', or 'Low' based on revenue thresholds.

    Thresholds:
    - High: > 3500
    - Medium: 2000 to 3500 (inclusive)
    - Low: < 2000

    Parameters:
    sales_amount (float): The revenue amount for a day.

    Returns:
    str: 'High', 'Medium', or 'Low'
    """
    if sales_amount > 3500:
        return 'High'
    elif sales_amount >= 2000:
        return 'Medium'
    else:
        return 'Low'

# 4. Implement a custom list-reordering function using Bubble Sort
def custom_sort_by_revenue(dataset, descending=True):
    """
    Sorts the dataset in-place based on daily sales revenue using the Bubble Sort algorithm.

    Parameters:
    dataset (list of dict): The list of dictionaries containing daily sales data.
    descending (bool): If True, sorts from highest to lowest revenue. Otherwise, lowest to highest.

    Returns:
    list of dict: The sorted dataset list.
    """
    n = len(dataset)
    # Temporary list to avoid changing the original input if desired, but we will sort in-place for pure sorting logic
    sorted_list = list(dataset)

    for i in range(n):
        for j in range(0, n - i - 1):
            rev1 = sorted_list[j]['quantity'] * sorted_list[j]['price_per_unit']
            rev2 = sorted_list[j+1]['quantity'] * sorted_list[j+1]['price_per_unit']

            if descending:
                if rev1 < rev2:
                    sorted_list[j], sorted_list[j+1] = sorted_list[j+1], sorted_list[j]
            else:
                if rev1 > rev2:
                    sorted_list[j], sorted_list[j+1] = sorted_list[j+1], sorted_list[j]
    return sorted_list

# Verify the implementations
print("=== Dataset Verification ===")
for row in wanjiku_dataset:
    daily_revenue = row['quantity'] * row['price_per_unit']
    classification = classify_day(daily_revenue)
    print(f"{row['day']}: {row['item']} - Revenue: {daily_revenue} KES, Class: {classification}")

total_rev = week_totals(wanjiku_dataset)
print(f"\nTotal Weekly Sales Revenue: {total_rev} KES")

print("\n=== Dataset Sorted by Revenue (Descending) ===")
sorted_data = custom_sort_by_revenue(wanjiku_dataset, descending=True)
for row in sorted_data:
    daily_revenue = row['quantity'] * row['price_per_unit']
    print(f"{row['day']}: {row['item']} - Revenue: {daily_revenue} KES")
```

## generate_performance_report

### Subtask:
Generate and print the structured 'Stall Performance Report' formatted with thousands separators.


**Reasoning**:
Generate the Stall Performance Report as specified in the subtask instructions, displaying the daily transaction breakdowns with classifications and formatting all currency values (KES) with thousands separators, followed by the weekly summary.



```python
def generate_stall_report(dataset):
    print("=" * 65)
    print("                  WANJIKU'S STALL PERFORMANCE REPORT")
    print("=" * 65)
    print(f"{'Day':<11} | {'Item':<20} | {'Qty':<4} | {'Price':<6} | {'Revenue (KES)':<13} | {'Class':<6}")
    print("-" * 65)

    for record in dataset:
        revenue = record['quantity'] * record['price_per_unit']
        classification = classify_day(revenue)
        revenue_formatted = f"{revenue:,.2f}"
        price_formatted = f"{record['price_per_unit']:,.0f}"
        print(f"{record['day']:<11} | {record['item']:<20} | {record['quantity']:<4} | {price_formatted:>6} | {revenue_formatted:>13} | {classification:<6}")

    print("-" * 65)
    total_revenue = week_totals(dataset)
    total_revenue_formatted = f"{total_revenue:,.2f}"
    print(f"{'TOTAL WEEKLY REVENUE:':<48} {total_revenue_formatted:>13} KES")
    print("=" * 65)

generate_stall_report(wanjiku_dataset)
```
